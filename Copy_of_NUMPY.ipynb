```python
import numpy as np
```

```python
# full(shape,value) ...to make array
a = np.full((1,3), 4)
print(a)
```

```python
# zeros(value)
a = np.zeros(1)
print(a)
```

```python
# identitymatrix......eye(2)
a = np.eye(3)
print(a)
```

```python
# shape (rows and columns)
d_arr = np.array([[2,2],[2,3],[8,3]])
print(d_arr.shape)
```

```python
# size ..total number of array-
a = np.array([[1,2,3],[1,2,3]])
print(a.size)
```

```python
# dimension 1d arr, 2d arr , 3d arr
a = np.array([1,2,2])
b = np.array([[1,2],[1,2],[1,2]])
c = np.array([[[[1,2],[1,2],[1,2]]]])
print(a.ndim)
print(c.ndim)

```

```python
#  to check type of the data
a = np.array([1,2,2])
print(a.dtype)
```

```python
# to convert the datatype
a = np.array([1,2.4,2])
print(a.astype(int))
```

```python
a = np.array([1,2,2,3,3,3,5,67,6])
print(sum(a))
print(min(a))
print(np.std(a))
print(np.var(a))
```

```python
# SLICING AND INDEXING
a = np.array([1,2,3,5,67,6])
a[4]
```

```python
# REVERSE THE ARRAY WITHOUT USING THE LOOP
a = np.array([10,20,30,50,60,70])
a[1:3]
a[::1]
# a[::-1] reverse
a[:3:2]

```

```python
# RESHAPING  ...dimension change krta h
a = np.array([1,2,3,5,67,6])
b = a.reshape(2,3)
print(b)
```

```python
#  multi dimension se one dimension m krta h  #revail # flatten
a = np.array([[1,2],[3,5],[67,6]])
print(a.flatten())
print(a.ravel())
```

```python
# insert the values (array , index , value)
a = np.array([10,20,30,50,60,70])
b = np.insert(a,2,4)
print(b)
```

```python
#  insert rows into the 2d array
a = np.array([[10,20],[30,50],[60,70]])
b = np.insert(a, 1,[2,3,34] ,axis =1)
print(b)
```

```python
# to concatenate the array
a = np.array([10,20,30,50,60,70])
b = np.array([10,20,30,50,60,70])
c = np.concatenate((a, b))
print(c)
```

```python
# BOARDCASTING IN NUMPY
a = np.array([1,2,3])
b = np.array([1,2,3])
print(a+b)
```

```python
a = np.array([[1,2,3] ,[1,2,3]])
a = np.array([1,2,3])
print(a+b)




```
